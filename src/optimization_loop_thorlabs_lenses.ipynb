# Optical architecture and étendue optique formalism.

In this notebook, we model the dual-path optical architecture designed to manipulate and detect a cloud of cold $\mathrm{}^{87}\mathrm{Rb}$ atoms. The setup couples two independent optical lines through the same physical space at the atomic plane: a trapping line that projects optical tweezers using a Spatial Light Modulator (SLM) at $\lambda_{\mathrm{trap}} = 850\text{ nm}$, and a fluorescence imaging line that collects the isotropic atomic emission at $\lambda_{\mathrm{fluo}} = 780\text{ nm}$. Fusing these paths into a single analytical framework allows us to understand how design choices in the illumination stage geometrically constrain the detection efficiency and resolution.

$$\text{SLM} \rightarrow \text{Relay telescope 1} \rightarrow \text{Objective 1} \rightarrow \text{Atoms} \rightarrow \text{Objective 2} \rightarrow \text{Relay telescope 2} \rightarrow \text{Tube lens} \rightarrow \text{qCMOS sensor}$$

To establish a rigorous framework for our exploration, we define the core physical constraints that govern the high-performance microscope objectives bounding the vacuum chamber. The Field of View $\left(\mathrm{FoV} \simeq 1.5\text{ mm}\right)$ defines the spatial range from the optical axis where geometric aberrations remain tightly bounded (typically quantified by a Strehl ratio $S \ge 0.8$). The Numerical Aperture $\left(\mathrm{NA} = n\sin \theta \simeq 0.65\right)$ represents the sine of the half-angle of the light cone, determining both the minimum waist of the traps and the resolving power of the fluorescence detection. Finally, the Working Distance ($\mathrm{WD} \simeq 23\text{ mm}$) imposes the strict physical clearance required between the mechanical edge of the objectives and the atomic vacuum chamber window.

### The role of focal length $f$ and pupil diameter $D$.

The choice of the objective's Effective Focal Length (EFL) $f$ sets off a direct tension between spatial angles and physical beam sizes across both subsystems. On the one hand, the displacement of an image point in the focal plane relates to the incidence angle by $x_{\mathrm{field}} = f \cdot \theta_{\mathrm{in}}$. To capture or project across the full extent of our field, the maximum tilt angle required to reach the edge of the $\mathrm{FoV}$ is given by:

$$\theta_{\mathrm{in}}^{\mathrm{max}} = \frac{\mathrm{FoV} / 2}{f}$$

From a fabrication standpoint, minimizing $\theta_{\mathrm{in}}^{\mathrm{max}}$ by choosing a large focal length $f$ is highly desirable because smaller propagation angles minimize geometric aberrations. However, the focal length simultaneously dictates the physical diameter $D$ of the beam entering or exiting the objective to achieve our targeted $\mathrm{NA}$:

$$D = 2 \cdot \mathrm{NA} \cdot f$$

For a fixed $\mathrm{NA}$, an overly large $f$ forces us to work with a massive clear aperture $D$. This places a severe burden on the upstream optical design, requiring expensive, large-diameter optics (such as the relay telescope lenses) to avoid clipping the trapping or fluorescence beams before they reach their destination. Our exploration will map out the viable window for $f$ under these geometric limits.

### Conservation of the étendue optique and system interplay.

Multiplying the beam diameter $D$ by the angular spread $\theta_{\mathrm{in}}^{\mathrm{max}}$ at the lens reveals a fundamental geometric constraint known as the **étendue optique** ($G$):

$$G = D \cdot \theta_{\mathrm{in}}^{\mathrm{max}} = (2 \cdot \mathrm{NA} \cdot f) \cdot \left(\frac{\mathrm{FoV} / 2}{f}\right) = \mathrm{FoV} \cdot \mathrm{NA}$$

In paraxial ray tracing, this principle is formally known as the Lagrange invariant (or Smith-Helmholtz invariant), expressed as $H = n \cdot y \cdot u = \mathrm{const}$. Named after the mathematician Joseph-Louis Lagrange during his foundational work on optical systems, it dictates that the product of the field height ($y$) and the marginal ray angle ($u$) is perfectly conserved throughout any ideal, lossless optical train. Because geometric optics preserves this phase-space volume, the focal length $f$ drops out entirely—it has absolutely no effect on this invariant quantity.

We can interpret this invariant through diffraction and information theory. The maximum number of independent spatial modes (or distinct optical traps) $N_{\mathrm{modes}}$ that can be packed into our field of view is determined by dividing the total field size by the minimum resolvable separation distance $\Delta x_{\mathrm{min}}$. Whether we define this threshold using the Gaussian waist radius ($\Delta x_{\mathrm{min}} = \omega_0=\frac{\lambda}{\pi \cdot \mathrm{NA}}$), the full beam diameter ($\Delta x_{\mathrm{min}} = 2\omega_0$), or the classical Rayleigh criterion ($\Delta x_{\mathrm{min}} = 0.61 \cdot \frac{\lambda}{\mathrm{NA}}$) for circular pupils, the structural scaling law remains invariant:

$$N_{\mathrm{modes}} =\frac{\mathrm{FoV}}{\Delta x_{\mathrm{min}}} \propto \frac{\mathrm{FoV} \cdot \mathrm{NA}}{\lambda}=\frac{G}{\lambda}$$

The étendue optique, normalized by the wavelength, defines the absolute information capacity of the system. It dictates how many independent spatial channels we can separate before diffraction causes neighboring optical traps to overlap or causes the light collected from adjacent atoms to irreversibly blend together, drawing a direct parallel to the Nyquist-Shannon sampling theorem in signal processing.

## Effective $\text{FoV}$ and $\text{NA}$ in the trapping line.

The operational workspace of a cold-atom trapping and imaging setup is fundamentally bounded by the conservation of the étendue optique throughout the multi-stage optical train, acting as a rigid information budget. At the At the source of the trapping line, the Spatial Light Modulator (SLM) introduces a strict geometric limit on the phase-engineering capacity due to its pixelated structure. The minimum spatial period required to deflect a laser beam without digital aliasing comprises exactly two pixels, a threshold established by the spatial Nyquist criterion where a continuous phase ramp must be sampled by at least two discrete points per period ($2p$, where $p = 12.5\ \mu\text{m}$ represents the physical pixel pitch). For an optical tweezers trapping wavelength of $\lambda_{\text{trap}} = 850\text{ nm}$, the maximum clean deflection angle in the first diffraction order is strictly constrained to $\theta_{\text{SLM}}^{\text{max}} = \arcsin\frac{\lambda_{\text{trap}}}{2p} \simeq 34\text{ mrad}$. By illuminating a fixed spatial diameter $D_{\text{SLM}} \simeq 12.35\text{ mm}$ on the modulator surface, the experimental configuration injects a definitive and invariant phase-space volume into the system, yielding a constant total étendue of $G_{\text{trap}} = D_{\text{SLM}} \cdot \theta_{\text{SLM}}^{\text{max}} \simeq 0.42\text{ mm}$. Because this volume is an optical invariant, it is perfectly preserved through the lossless relay optics down to the atomic plane, enforcing a continuous hyperbolic coupling that dictates the effective Field of View ($\text{FoV}_{\text{eff}}$) and the effective numerical aperture ($\text{NA}_{\text{eff}}$) at the atoms via:

$$G_{\text{trap}} = D_{\text{SLM}} \cdot \theta_{\text{SLM}}^{\text{max}} = \text{FoV}_{\text{eff}} \cdot \text{NA}_{\text{eff}}$$

Physically, one navigates along this continuous hyperbolic frontier by **tuning the magnification ($M$) of the relay telescope positioned downstream of the SLM (between the SLM and the objective)**. Mathematically, this transformation maps the coordinates from the SLM plane to the atomic plane through the objective focal length ($f_{\text{obj}} = 14\text{ mm}$), yielding the explicit linear scaling for the field size and the inverse scaling for the resolution:

$$\text{FoV}_{\text{eff}} = \frac{2 \cdot f_{\text{obj}} \cdot \theta_{\text{SLM}}^{\text{max}}}{M}; \quad \text{NA}_{\text{eff}} \simeq \frac{D_{\text{SLM}} \cdot M}{2 \cdot f_{\text{obj}}}$$

This explicit dependence dictates a definitive trend along the invariant hyperbola depending on the chosen optical scale. Increasing the telescope magnification ($M > 1$) shifts the operating point toward the high-resolution regime; the beam expands to fill a larger fraction of the objective's pupil, which inherently compresses the angular deflection range. In the asymptotic limit of this trend, where the beam completely fills the pupil's nominal capacity, the system locks into a maximum resolution of $\text{NA}_{\text{eff}} = 0.65$, confining the atomic array to a minimum central Field of View of $\text{FoV}_{\text{eff}} \simeq 0.64\text{ mm}$. Conversely, decreasing the magnification toward a demagnifying regime ($M < 1$) drives the system in the opposite direction along the curve. This trend compresses the beam into a narrower pencil of rays at the pupil plane, providing the geometric clearance necessary to transmit progressively wider diffraction angles without clipping the lens edges. At the boundary of this wide-field trend, the trapping landscape expands to its maximum physical limit of $\text{FoV}_{\text{eff}} \simeq 1.5\text{ mm}$, at the structural cost of lowering the convergence angle of the traps to an asymptotic floor of $\text{NA}_{\text{eff}} \simeq 0.28$.

The following plot illustrates this hyperbola of accessible values within the system's phase space, mapping the continuous trade-off between the achievable atomic array field size and the local numerical aperture of the trapping channels.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from collections import Counter

# global configuration and constants
MAIN_PURPLE = "#4A154B"
COLOR_CURVE = "#912491"
BG_LIGHT_PURPLE = "#f4f0fa"
GRID_COLOR = "#b0a8b9"

# configure global matplotlib parameters for consistent figure styling
plt.rcParams.update({
    'axes.labelweight': 'bold',
    'axes.titleweight': 'bold',
    'figure.facecolor': '#ffffff',
    'axes.facecolor': BG_LIGHT_PURPLE,
    'grid.color': GRID_COLOR,
    'grid.linestyle': '--',
    'grid.alpha': 0.4
})

# physical parameters of the optical trapping system
WAVELENGTH_TRAP = 850e-9  # wavelength of the trapping laser in meters
OBJECTIVE_NA = 0.65  # numerical aperture of the objective lens
MAX_PHYSICAL_FOV = 1.5e-3  # maximum physical field of view in meters
OBJECTIVE_EFL = 17.5e-3  # effective focal length of the objective lens in meters
PIXEL_PITCH = 12.5e-6  # spatial light modulator pixel pitch in meters
D_SLM = 12.35e-3  # active clear aperture diameter of the slm in meters

# trapping line parametric calculations
# calculate maximum deflection angle allowed by slm pixel pitch based on nyquist limit
theta_max_slm = np.arcsin(WAVELENGTH_TRAP / (2.0 * PIXEL_PITCH))
# calculate invariant optical etendue for the trapping line
etendue_G_trap = D_SLM * theta_max_slm

# calculate maximum 1d spatial modes invariant representing resolving capacity
# N_modes = (FoV * NA) / lambda = G_trap / lambda
n_modes_1d = etendue_G_trap / WAVELENGTH_TRAP

# calculate boundary condition for regime 1: high-resolution mode (full NA utilization)
m_high_res = (OBJECTIVE_NA * 2.0 * OBJECTIVE_EFL) / D_SLM
fov_high_res = (2.0 * OBJECTIVE_EFL * theta_max_slm) / m_high_res
waist_high_res = WAVELENGTH_TRAP / (np.pi * OBJECTIVE_NA)

# calculate boundary condition for regime 2: wide-field mode (full FoV utilization)
m_wide_field = (2.0 * OBJECTIVE_EFL * theta_max_slm) / MAX_PHYSICAL_FOV
na_wide_field = (D_SLM * m_wide_field) / (2.0 * OBJECTIVE_EFL)
waist_wide_field = WAVELENGTH_TRAP / (np.pi * na_wide_field)

# print calculated system invariants and regime limits to console
print("=== Trapping line boundary conditions ===")
print(f"Immutable trapping étendue optique (G_trap): {etendue_G_trap * 1e3:.4f} mm")
print(f"Maximum deflection angle at SLM:    {theta_max_slm * 1e3:.2f} mrad")
print(f"Information capacity (N_modes 1D):  {int(np.round(n_modes_1d))} traps\n")

print("=== Regime 1. High-resolution (full NA) ===")
print(f"Required telescope magnification:    {m_high_res:.2f}x")
print(f"Effective Field of View (FoV_eff):   {fov_high_res * 1e3:.2f} mm")
print(f"Effective Numerical Aperture (NA):   {OBJECTIVE_NA:.2f}")
print(f"Individual optical trap waist:       {waist_high_res * 1e6:.3f} um\n")

print("=== Regime 2. Wide-field (full FoV) ===")
print(f"Required telescope magnification:    {m_wide_field:.2f}x")
print(f"Effective Field of View (FoV_eff):   {MAX_PHYSICAL_FOV * 1e3:.2f} mm")
print(f"Effective Numerical Aperture (NA):   {na_wide_field:.2f}")
print(f"Individual optical trap waist:       {waist_wide_field * 1e6:.3f} um")

# phase space sweep and plotting
# sweep relay telescope magnification values across operational range
m_sweep = np.linspace(0.3, 3.0, 1000)

# compute effective field of view and numerical aperture across magnification sweep
fov_eff_sweep = (2.0 * OBJECTIVE_EFL * theta_max_slm) / m_sweep
na_eff_sweep = (D_SLM * m_sweep) / (2.0 * OBJECTIVE_EFL)
fov_eff_mm = fov_eff_sweep * 1e3

# filter parameters within physical constraints of objective na and maximum fov
valid_mask = (na_eff_sweep <= OBJECTIVE_NA) & (fov_eff_mm <= (MAX_PHYSICAL_FOV * 1e3))
fov_plot = fov_eff_mm[valid_mask]
na_plot = na_eff_sweep[valid_mask]

# initialize primary plot figure
fig, ax = plt.subplots(figsize=(8.5, 5.8))

# plot trade-off curve between field of view and numerical aperture
ax.plot(fov_plot, na_plot, color=COLOR_CURVE, lw=3)

# safe coordinate transformation functions for secondary axes avoiding division by zero
def fov_to_mag(x):
    # map field of view in millimeters to telescope magnification
    x_safe = np.where(x <= 0, np.nan, x)
    with np.errstate(divide='ignore', invalid='ignore'):
        return (2.0 * OBJECTIVE_EFL * theta_max_slm) / (x_safe * 1e-3)

def mag_to_fov(x):
    # map telescope magnification to field of view in millimeters
    x_safe = np.where(x <= 0, np.nan, x)
    with np.errstate(divide='ignore', invalid='ignore'):
        return ((2.0 * OBJECTIVE_EFL * theta_max_slm) / x_safe) * 1e3

def na_to_waist(x):
    # map numerical aperture to optical trap beam waist in micrometers
    x_safe = np.where(x <= 0, np.nan, x)
    with np.errstate(divide='ignore', invalid='ignore'):
        return (WAVELENGTH_TRAP * 1e6) / (np.pi * x_safe)

def waist_to_na(x):
    # map optical trap beam waist in micrometers to numerical aperture
    x_safe = np.where(x <= 0, np.nan, x)
    with np.errstate(divide='ignore', invalid='ignore'):
        return (WAVELENGTH_TRAP * 1e6) / (np.pi * x_safe)

# create secondary top axis for telescope magnification
ax_top = ax.secondary_xaxis('top', functions=(fov_to_mag, mag_to_fov))
ax_top.set_xticks([2.5, 2.0, 1.75, 1.5, 1.25, 1.0, 0.75])
ax_top.set_xlabel(r"Relay telescope magnification ($\mathbf{M}$)", fontsize=18, color=MAIN_PURPLE, weight="bold", labelpad=12)

# create secondary right axis for individual trap beam waist
ax_right = ax.secondary_yaxis('right', functions=(na_to_waist, waist_to_na))
ax_right.set_ylabel(r"Individual optical trap waist ($\mathbf{\boldsymbol{\omega}_0}$) [$\mathbf{\mu m}$]", fontsize=18, color=MAIN_PURPLE, weight="bold", labelpad=12)

# mark and annotate high-resolution regime operating point
ax.scatter(fov_high_res * 1e3, OBJECTIVE_NA, color=MAIN_PURPLE, s=25, zorder=5)
ax.annotate(f"High-resolution regime\nNA = {OBJECTIVE_NA:.2f}, FoV = {fov_high_res * 1e3:.2f} mm\n(M = {m_high_res:.2f}x)", 
            (fov_high_res * 1e3, OBJECTIVE_NA), textcoords="offset points", xytext=(30, -30), ha="left", va="bottom",
            weight="bold", color=MAIN_PURPLE, fontsize=15,
            arrowprops=dict(arrowstyle="->", color=MAIN_PURPLE, lw=1.2))

# mark and annotate wide-field regime operating point
ax.scatter(MAX_PHYSICAL_FOV * 1e3, na_wide_field, color=MAIN_PURPLE, s=25, zorder=5)
ax.annotate(f"Wide-field regime\nNA = {na_wide_field:.2f}, FoV = {MAX_PHYSICAL_FOV * 1e3:.2f} mm\n(M = {m_wide_field:.2f}x)", 
            (MAX_PHYSICAL_FOV * 1e3, na_wide_field), textcoords="offset points", xytext=(75, 40), ha="right", va="bottom",
            weight="bold", color=MAIN_PURPLE, fontsize=15,
            arrowprops=dict(arrowstyle="->", color=MAIN_PURPLE, lw=1.2))

# overlay textbox summarizing etendue and spatial mode invariant equations
info_box_text = (
    r"$\mathbf{G_{trap} = D_{SLM} \cdot \theta_{SLM}^{max} \simeq 0.42\ mm}$" + "\n" +
    r"$\mathbf{N_{modes} \simeq \frac{G_{trap}}{\lambda_{trap}} \simeq 495\ 1D\ modes}$"
)
ax.text(0.04, 0.06, info_box_text, transform=ax.transAxes, fontsize=16,
        color=MAIN_PURPLE, weight="bold", va="bottom", ha="left",
        bbox=dict(boxstyle="round,pad=0.5", edgecolor=MAIN_PURPLE, facecolor=BG_LIGHT_PURPLE, alpha=0.9, lw=1.2))

# configure primary axes labels
ax.set_xlabel(r"Effective Field of View ($\mathbf{FoV_{eff}}$) [mm]", fontsize=18, color=MAIN_PURPLE, weight="bold")
ax.set_ylabel(r"Effective Numerical Aperture ($\mathbf{NA_{eff}}$)", fontsize=18, color=MAIN_PURPLE, weight="bold")

# set primary plot display limits
ax.set_xlim(0.6, 1.7)
ax.set_ylim(0.2, 0.7)

# configure grid display properties
ax.grid(True, which="both", linestyle=":", alpha=0.5)

# format axis lines and tick parameters across all axes
for axis in [ax, ax_top, ax_right]:
    axis.tick_params(colors=MAIN_PURPLE, labelsize=18, width=1.5)
    for spine in axis.spines.values():
        spine.set_edgecolor(MAIN_PURPLE)
        spine.set_linewidth(1.5)

# apply bold formatting to all axis tick labels
for label in ax.get_xticklabels() + ax.get_yticklabels():
    label.set_weight('bold')
for label in ax_top.get_xticklabels():
    label.set_weight('bold')
for label in ax_right.get_yticklabels():
    label.set_weight('bold')

# adjust subplot padding and show plot
plt.tight_layout()
plt.show()
```

## Fluorescence line.

The fluorescence imaging line operates in the inverse direction of the trapping mechanism. Here, the source consists of $\mathrm{}^{87}\mathrm{Rb}$ atoms that, upon excitation, emit photons isotropically ($4\pi\text{ sr}$) at the rubidium $D_2$ transition wavelength $\left(\lambda_{\mathrm{fluo}} = 780\text{ nm}\right)$. In our architecture, the primary objective of this line is not to spatially resolve the internal geometric profile of each optical tweezer, but rather to perform high-fidelity binary classification: establishing with maximum statistical confidence whether a specific trapping site is occupied by a single atom or empty.

To achieve this metadata extraction, the system must collect as many scattered photons as possible within the short coherence window of the experimental cycle, maximizing the signal-to-noise ratio (SNR) against the readout noise of the sensor. Therefore, Objective 2 must operate at its full nominal capacity, capturing a complete numerical aperture of $\mathrm{NA} = 0.65$ across the maximum designated field of view ($\mathrm{FoV} = 1.5\text{ mm}$). This high-$\mathrm{NA}$ collector injects a maximum phase-space volume into the downstream imaging train, yielding an unalterable baseline fluorescence étendue of:

$$G_{\mathrm{fluo}} = \mathrm{FoV} \cdot \mathrm{NA} = 1.5\text{ mm} \cdot 0.65 = 0.975\text{ mm}$$

Because geometric optics preserves this phase-space volume $\left(H = n \cdot y \cdot u = \text{const}\right)$, our optical design task reduces to a matching problem: we must transport and scale $G_{\mathrm{fluo}}$ through a combination of commercial lenses so that the photon flux from the entire atomic cloud is efficiently mapped onto the physical dimensions of our qCMOS sensor (Hamamatsu ORCA-Quest 2, featuring a pixel pitch of $p_{\mathrm{sensor}} = 4.6\ \mu\text{m}$).

### Magnification bounding.

The lower bound of our magnification architecture is dictated by the absolute operational requirement of single-site distinguishability. Given a physical lattice spacing or minimum separation distance between adjacent trapping sites, denoted as $\Delta x_{\mathrm{atoms}}$, the optical line must project these discrete emitter centers onto the qCMOS sensor with sufficient spatial clearance. To ensure that under no circumstance the collected fluorescence from two neighboring atoms bleeds into a single pixel coordinate, the projected separation between their geometric centers must strictly exceed the physical pixel pitch $p_{\mathrm{sensor}}$, factoring in the operational diffraction blur. This hard geometric constraint establishes a baseline magnification threshold:

$$M_{\mathrm{min}} > \frac{p_{\mathrm{sensor}}}{\Delta x_{\mathrm{atoms}}}$$

Conversely, the upper bound remains governed by the physical boundaries of the imaging hardware. To avoid spatial truncation of our data channels, the full extended Field of View must be completely contained within the active area of the sensor. For a qCMOS matrix defined by a minimum clearing dimension $y_{\mathrm{sensor}}$, the maximum allowable magnification before geometric clipping occurs is restricted to:

$$M_{\mathrm{max}} < \frac{\min\{x_{\mathrm{sensor}}, \ y_{\mathrm{sensor}}\}}{\mathrm{FoV}}=\frac{y_{\mathrm{sensor}}}{\mathrm{FoV}}$$

This dual-boundary framework defines a strict operational window $\left(M_{\mathrm{min}}, M_{\mathrm{max}}\right)$. Any commercial lens combination must position the system magnification within these precise coordinates to balance discrete spatial isolation against full-field information retention.

```python
# Minimal separation between adjacent atomic traps in the object plane
ATOMS_SEPARATION, PIXEL_SIZE = 10.0e-6, 4.6e-6
SENSOR_HEIGHT, SENSOR_WIDTH = 18.841e-3, 10.598e-3

# Restrictive magnification limits based on pixel isolation and sensor size
# Minimum magnification required to isolate single atomic traps across adjacent sensor pixels
mag_min = PIXEL_SIZE / ATOMS_SEPARATION
# Maximum magnification allowed to enclose full physical fov within active sensor dimensions
mag_max_full_FoV = min(SENSOR_HEIGHT, SENSOR_WIDTH) / MAX_PHYSICAL_FOV
# Maximum magnification allowed to enclose full high-resolution fov within active sensor dimensions
mag_max_full_NA = min(SENSOR_HEIGHT, SENSOR_WIDTH) / fov_high_res
# Target magnification mapping wide-field optical trap waist to single sensor pixel size
mag_targ = PIXEL_SIZE / waist_wide_field

# Calculate the precise pixel spacing separating adjacent atomic traps at target magnification
pixel_separation_at_target = (ATOMS_SEPARATION * mag_targ) / PIXEL_SIZE

# Print magnification bounds and trap spacing results to console
print(f"=== Magnification bounds analysis ===")
print(f"Minimum magnification for pixel isolation: {mag_min:.3f}x")
print(f"Maximum magnification for full FoV enclosure (trapping system): {mag_max_full_FoV}x")
print(f"Maximum magnification for full NA enclosure (trapping system): {mag_max_full_NA}x")
print(f"Target magnification for 1 px to 1 atom: {mag_targ:.3f}x")
print(f"Average pixel separation between adjacent atomic traps: {pixel_separation_at_target:.2f} pixels")
```

### Geometric trade-offs.

To position the total system magnification within our operational boundaries ($M_{\mathrm{min}} < M_{\mathrm{total}} < M_{\mathrm{max}}$), the imaging line cannot rely on a single tube lens positioned immediately after the second objective. Physical constraints, such as the strict mechanical clearance required around the vacuum chamber and the mandatory insertion of narrow-band dielectric filters to isolate the $780\text{ nm}$ fluorescence from the intense $850\text{ nm}$ trapping background, necessitate a multi-stage optical architecture. By introducing an intermediate relay telescope consisting of two lenses ($f_{\mathrm{relay1}}$ and $f_{\mathrm{relay2}}$) configured in a strict afocal 4f arrangement, the system scales both the collimated on-axis beam waist and the off-axis angular field distribution. 

Crucially, this architecture exhibits a decoupled behavior between the physical beam footprint and the actual spatial scale of the image projected onto the sensor. While the nominal collimated beam diameter scales proportionally with the relay ratio, the Lagrange invariant (conservation of the *étendue optique*) dictates that the spatial compression of a beam must be accompanied by a reciprocal expansion of its angular field distribution. 

To formally derive the physical height of the image at the camera plane ($y_{\mathrm{sensor}}$), we track the **paraxial evolution of the *chief ray***—defined at the atomic object plane ($z=0$) with an initial off-axis boundary height $y_0$ and an on-axis parallel propagation angle $\theta_0 = 0$. Using the standard ABCD thin-lens and free-space transfer matrices, the sequential propagation steps develop as follows:

1. **Propagation to the objective:** The ray travels a free space equal to the objective focal length $f_{\mathrm{obj}}$, reaching the lens at a height $y_1 = y_0 + \theta_0 \cdot f_{\mathrm{obj}} = y_0$. The objective refracts the ray, modifying its angle to:
   $$\theta_1 = \theta_0 - \frac{y_1}{f_{\mathrm{obj}}} = -\frac{y_0}{f_{\mathrm{obj}}}$$

2. **Propagation to relay lens 1:** In a strict afocal configuration, the separation between the objective and the first relay lens is set to the sum of their focal lengths ($f_{\mathrm{obj}} + f_{\mathrm{relay1}}$). The ray height upon impacting Relay 1 becomes:
   $$y_2 = y_1 + \theta_1 \cdot (f_{\mathrm{obj}} + f_{\mathrm{relay1}}) = y_0 + \left(-\frac{y_0}{f_{\mathrm{obj}}}\right)(f_{\mathrm{obj}} + f_{\mathrm{relay1}}) = -y_0 \cdot \frac{f_{\mathrm{relay1}}}{f_{\mathrm{obj}}}$$
   Refraction through Relay 1 alters the angular trajectory to:
   $$\theta_2 = \theta_1 - \frac{y_2}{f_{\mathrm{relay1}}} = -\frac{y_0}{f_{\mathrm{obj}}} - \frac{-y_0 \cdot \frac{f_{\mathrm{relay1}}}{f_{\mathrm{obj}}}}{f_{\mathrm{relay1}}} = 0$$
   As expected for a 4f system, the chief ray becomes perfectly parallel to the optical axis in the intermediate focal plane.

3. **Propagation to relay lens 2:** The two relay lenses share a common focal plane, separated by a distance $f_{\mathrm{relay1}} + f_{\mathrm{relay2}}$. Since $\theta_2 = 0$, the ray maintains its height during this interval:
   $$y_3 = y_2 + \theta_2 \cdot (f_{\mathrm{relay1}} + f_{\mathrm{relay2}}) = -y_0 \cdot \frac{f_{\mathrm{relay1}}}{f_{\mathrm{obj}}}$$
   Upon traversing Relay 2, the new refraction angle is injected as:
   $$\theta_3 = \theta_2 - \frac{y_3}{f_{\mathrm{relay2}}} = 0 - \frac{-y_0 \cdot \frac{f_{\mathrm{relay1}}}{f_{\mathrm{obj}}}}{f_{\mathrm{relay2}}} = \frac{y_0 \cdot f_{\mathrm{relay1}}}{f_{\mathrm{obj}} \cdot f_{\mathrm{relay2}}}$$
   Utilizing the nominal relay magnification definition ($M_{\mathrm{relay}} = f_{\mathrm{relay2}}/f_{\mathrm{relay1}}$), this angular deviation simplifies to:
   $$\theta_3 = \frac{y_0}{f_{\mathrm{obj}} \cdot M_{\mathrm{relay}}}$$

4. **Propagation through the tube lens to the qCMOS sensor:** The ray travels the final stages through the tube lens ($f_{\mathrm{relay2}} + f_{\mathrm{tube}}$) and undergoes its final refraction. In a paraxial system, a tube lens maps any incoming angular deviation $\theta_{\mathrm{in}}$ directly into a physical displacement at its back focal plane via $y_{\mathrm{final}} = -\theta_{\mathrm{in}} \cdot f_{\mathrm{tube}}$. Substituting our refracted angle $\theta_3$ yields the exact physical height on the sensor:
   $$y_{\mathrm{sensor}} = -\theta_3 \cdot f_{\mathrm{tube}} = -y_0 \cdot \frac{f_{\mathrm{tube}}}{f_{\mathrm{obj}} \cdot M_{\mathrm{relay}}}$$

By extracting the absolute field magnification factor ($M_{\mathrm{field}} = \lvert \frac{y_{\mathrm{sensor}}}{y_0} \rvert$), the strict analytical relationship governing the image scale on the qCMOS is revealed:

$$M_{\mathrm{field}} = \frac{f_{\mathrm{tube}}}{f_{\mathrm{obj}} \cdot M_{\mathrm{relay}}} = \frac{f_{\mathrm{tube}}}{f_{\mathrm{obj}}} \cdot \frac{f_{\mathrm{relay1}}}{f_{\mathrm{relay2}}}\Leftrightarrow f_{\mathrm{tube}}=M_{\mathrm{field}}\cdot M_{\mathrm{relay}} \cdot f_{\mathrm{obj}}$$

Tuning the magnification of the relay stage $\left(M_{\mathrm{relay}}\right)$ modifies the structural demands of the downstream optics. The core beam diameter exiting the objective $\left(D_{\mathrm{pupil}} = 2 \cdot \mathrm{NA} \cdot f_{\mathrm{obj}} \simeq 18.2\text{ mm}\right)$ scales directly with the relay ratio to become $D_{\mathrm{tube}} = D_{\mathrm{pupil}} \cdot M_{\mathrm{relay}}$ at the entrance of the tube lens. This coupling introduces a **severe geometric trade-off** when selecting commercial catalog components. Opting for a magnifying relay ($M_{\mathrm{relay}} > 1$) shifts the beam expansion burden onto the intermediate stage, which compresses the final field image on the sensor but forces the physical beam diameter $D_{\mathrm{tube}}$ to expand. This expansion can easily **exceed the Clear Aperture** $\left(\mathrm{CA}\right)$ of standard commercial optics (typically $\mathrm{CA} \simeq 0.9\varnothing$ for Thorlabs lenses), causing severe geometric clipping of the peripheral rays and heavy vignetting at the lens boundaries.

Conversely, designing a demagnifying relay $\left(M_{\mathrm{relay}} < 1\right)$ compresses the core beam cross-section, providing the necessary clearance to utilize compact, standard 1-inch optics down the line without clipping the lens frames. However, as required by the conservation of the phase-space volume, this spatial compression forces the angular spread of the peripheral field rays to expand inversely: $\theta_{\mathrm{out}} = \frac{\theta_{\mathrm{in}}}{M_{\mathrm{relay}}}$. Consequently, the *chief ray* originating from the edge of the field of view diverges rapidly from the optical axis during free-space propagation. This steep angular deviation drastically scales up the final image size on the qCMOS sensor, threatening to push the edge-most atomic signals outside the active vertical height of the chip. Furthermore, highly divergent rays striking downstream dielectric interference filters at steep angles can induce significant geometric aberrations and shift the central passband of the coatings, which degrades the global photon collection efficiency.

Beyond beam diameters and angular field expansions, the choices of individual focal lengths are bound by a strict laboratory space budget. In a realistic optomechanical assembly, the total physical track length $\left(L_{\mathrm{total}}\right)$ from the atomic plane inside the vacuum chamber to the qCMOS sensor cannot be modeled merely as the sum of ideal thin-lens focal distances. Instead, the spatial envelope is bounded from below by a strict geometric clearance that incorporates the objective's working distance $\left(\mathrm{WD} = 23\text{ mm}\right)$ and its massive physical housing length $\left(L_{\mathrm{obj}}\right)$. Treating the multi-element high-$\mathrm{NA}$ objective as a structural black box, a robust lower bound for the physical footprint of the imaging line is established as:

$$L_{\mathrm{total}} \ge \mathrm{WD} + L_{\mathrm{obj}} + 2 \cdot f_{\mathrm{relay1}} + 2 \cdot f_{\mathrm{relay2}} + f_{\mathrm{tube}}$$

This physical dimension imposes a hard architectural envelope on our design. On the lower end, an overly compressed setup $\left(L_{\mathrm{total}} < 30\text{ cm}\right)$ compromises ergonomics, leaving insufficient physical clearance to safely mount optomechanics, mirror mounts, or multi-axis filter wheels on the optical table. On the upper end, a bloated optical train $\left(L_{\mathrm{total}} > 120\text{ cm}\right)$ is highly susceptible to structural instability, thermal drifts, and mechanical vibrations, which can introduce pixel-scale pointing noise on the sensor. Consequently, the optimal Thorlabs lens combination must sit within a multi-dimensional intersection where strict field magnification, clear aperture thresholds, and laboratory track length are satisfied simultaneously.

### Optimization loop for commercial Thorlabs lenses.

To map these physical trade-offs into an automated selection tool, the numerical algorithm evaluates every mathematically possible triplet $(f_{\mathrm{relay1}}, f_{\mathrm{relay2}}, f_{\mathrm{tube}})$ extracted from the commercial Thorlabs inventory. A series of strict physical, geometric, and structural filters are applied in a cascading sequence to discard invalid combinations before populating the final procurement data frame:

* **Filter 1. Magnification bounding.**
  The total magnification of the system ($M_{\mathrm{total}}$) must fall strictly within the pre-defined target windows required for optimal spatial tracking:
  $$M_{\mathrm{min}} \le M_{\mathrm{total}} \le M_{\mathrm{max}}$$
  If a target magnification ($M_{\mathrm{target}}$) and a specific tolerance are defined, triplets that deviate beyond this boundary are immediately discarded.

* **Filter 2. Laboratory space and rail constraints.**
  The total optomechanical length ($L_{\mathrm{total}}$), accounting for the objective housing and the 4f telescope spacing, is bounded to match table layout feasibility and high-focal setups:
  $$50\text{ cm} \le L_{\mathrm{total}} \le 200\text{ cm}$$
  Combinations below $50\text{ cm}$ are excluded due to mechanical mounting clashes between optic mounts, while setups exceeding $200\text{ cm}$ ($2.0\text{ m}$) are discarded to prevent excessive table footprint and susceptibility to thermal/vibrational pointing drifts.

* **Filter 3. Collimated chief ray angular deviation.**
  To preserve paraxial behavior and minimize geometric aberrations across the field of view, the propagation angle of the chief ray originating from the extreme FoV edge ($\theta_{\mathrm{coll}}$) is evaluated strictly within the collimated beam spaces—specifically between the objective and the first relay lens ($f_{\mathrm{relay1}}$), and between the second relay lens ($f_{\mathrm{relay2}}$) and the tube lens ($f_{\mathrm{tube}}$). The chief ray tilt relative to the optical axis must comply with:
  $$\lvert \theta_{\mathrm{coll}} \rvert \le 5 ^\circ$$
  Exceeding this threshold triggers immediate rejection of the triplet to prevent high-order off-axis degradation (such as coma and astigmatism).

* **Filter 4. Outer diameter fill factor.**
  This filter prevents peripheral beam clipping and edge aberration degradation by limiting the total beam footprint ($D_{\mathrm{real}}$) as a fraction of the lens physical outer diameter ($\varnothing_{\mathrm{lens}}$), rather than its clear aperture. Accounting for field ray height ($y$), $D_{\mathrm{real}} = D_{\mathrm{beam}} + 2.0 \cdot |y|$ must satisfy:
  $$D_{\mathrm{real}} \le \text{Filling Factor} \cdot \varnothing_{\mathrm{lens}}$$
  The maximum allowable ratio is tailored to the lens topology:
  $$\text{Filling Factor} = \begin{cases} 30\% & \text{for simple planoconvex singlets} \\ 45\text{--}55\% & \text{for achromatic doublets} \\ 90\% & \text{for aspheric lenses} \end{cases}$$

* **Filter 5. Minimum focal length floor.**
  To maintain generous working distances, lower high-order aberrations, and leave ample optomechanical clearance for filter wheels, shutters, and motorized stages, a strict lower limit is applied across all individual elements in the triplet:
  $$f_{\mathrm{relay1}},\, f_{\mathrm{relay2}},\, f_{\mathrm{tube}} \ge 100\text{ mm}$$

```python
def trace_ray(y, theta, dist, focal):
    # Calculate the ray height at the next optical plane
    y_next = y + dist * theta
    # Apply thin lens refraction to update the ray angle
    theta_next = theta - (y_next / focal)
    return y_next, theta_next


def check_lens_procurement(
    focal_m, required_diameter_m, allowed_footprint_m, catalog_dict
):
    # Default state when the required clear aperture exceeds allowed limits
    min_size = "Oversized"
    size_order = ['0.5"', '1.0"', '1.5"', '2.0"', '3.0"', '4.0"']

    # Determine the smallest mechanical diameter that accommodates the beam
    for size in size_order:
        if (
            size in allowed_footprint_m
            and required_diameter_m <= allowed_footprint_m[size]
        ):
            min_size = size
            break

    # Early return if no standard diameter is large enough
    if min_size == "Oversized":
        return False, "Oversized"

    # Restrict search to diameter options equal to or larger than min_size
    allowed_sizes = size_order[size_order.index(min_size) :]
    focal_mm_nominal = round(focal_m * 1e3, 1)

    # Search commercial catalog for a focal length match within tolerance
    for commercial_size in allowed_sizes:
        if commercial_size in catalog_dict:
            if any(
                abs(f - focal_mm_nominal) < 0.2
                for f in catalog_dict[commercial_size]
            ):
                return True, commercial_size

    return False, "Not Available"


def get_lens_price(model_dict, size, focal):
    # Convert target focal length to millimeters for matching
    focal_mm = round(focal * 1e3, 1)
    # Search specs dictionary for matching size and focal length
    for specs in model_dict.values():
        if specs["size"] == size and abs(specs["focal"] - focal_mm) < 0.2:
            return specs["price"]
    return 0.0


def optimize_imaging_line(
    catalog_dict,
    specs_dict,
    lens_type="singlet",
    min_magnification=mag_min,
    max_magnification=mag_max_full_NA,
    objective_efl=OBJECTIVE_EFL,
    objective_na=OBJECTIVE_NA,
    min_track_m=0.50,
    max_track_m=2.00,
    min_focal_m=0.100,
    max_collimated_angle_deg=5.0,
    m_target=mag_targ,
    m_tolerance=None,
    symmetric_relay=False,
    max_physical_fov=MAX_PHYSICAL_FOV,
    fill_factor_limit_singlets=0.3,
    fill_factor_limit_doublets=0.72,
    fill_factor_limit_aspherics=0.9,
):
    # Assign maximum clear aperture fill factor based on lens technology
    if lens_type == "singlet":
        fill_factor_limit = fill_factor_limit_singlets
    elif lens_type == "doublet":
        fill_factor_limit = fill_factor_limit_doublets
    else:
        fill_factor_limit = fill_factor_limit_aspherics

    # Standard mechanical outer diameters converted to meters
    physical_diameters_m = {
        '0.5"': 12.7 * 1e-3,
        '1.0"': 25.4 * 1e-3,
        '1.5"': 38.1 * 1e-3,
        '2.0"': 50.8 * 1e-3,
        '3.0"': 76.2 * 1e-3,
        '4.0"': 101.6 * 1e-3,
    }

    # Compute maximum allowed clear aperture per mechanical footprint
    allowed_footprint_m = {
        size: d * fill_factor_limit for size, d in physical_diameters_m.items()
    }

    # Consolidate and extract unique focal lengths available in catalog
    all_focals_mm = []
    for focal_list in catalog_dict.values():
        all_focals_mm.extend(focal_list)
    unique_focals_m = np.unique(np.array(all_focals_mm)) * 1e-3

    # Define system constants and optical parameters
    f_obj = objective_efl
    y_edge_fov = max_physical_fov / 2.0
    max_angle_rad = np.deg2rad(max_collimated_angle_deg)
    theta_marginal = np.arcsin(objective_na)

    # Initialize tracking variables for candidate evaluation
    valid_combinations = []
    total_evaluated = 0
    rejection_reasons = Counter()

    # Iterate over available focal lengths for first relay lens
    for f_r1 in unique_focals_m:
        if f_r1 < min_focal_m:
            continue

        # Restrict second relay lens if symmetric configuration is requested
        f_r2_set = [f_r1] if symmetric_relay else unique_focals_m
        for f_r2 in f_r2_set:
            if f_r2 < min_focal_m:
                continue

            # Iterate over available focal lengths for tube lens
            for f_tube in unique_focals_m:
                if f_tube < min_focal_m:
                    continue

                total_evaluated += 1

                # Calculate overall optical magnification
                m_relay = f_r2 / f_r1
                m_total = (f_tube / f_obj) / m_relay

                # Filter candidates by global magnification limits
                if (
                    min_magnification is not None
                    and m_total < min_magnification
                ):
                    rejection_reasons[
                        "Magnification below lower limit"
                    ] += 1
                    continue
                elif (
                    max_magnification is not None
                    and m_total > max_magnification
                ):
                    rejection_reasons[
                        "Magnification above upper limit"
                    ] += 1
                    continue

                # Filter candidates by target magnification tolerance
                if m_tolerance is not None and m_target is not None:
                    if np.abs(m_total - m_target) > m_tolerance:
                        rejection_reasons[
                            "Magnification target tolerance exceeded"
                        ] += 1
                        continue

                # Calculate total system track length
                l_track = f_obj + (2.0 * f_r1) + (2.0 * f_r2) + f_tube

                # Filter candidates by track length constraints
                if l_track < min_track_m:
                    rejection_reasons["Track length below lower limit"] += 1
                    continue
                elif l_track > max_track_m:
                    rejection_reasons["Track length above upper limit"] += 1
                    continue

                # Trace chief ray through the system for field edge
                theta0_c = -y_edge_fov / f_obj
                y_obj_c, theta_obj_c = trace_ray(
                    y_edge_fov, theta0_c, f_obj, f_obj
                )

                # Propagate chief ray across pupil-conjugated relay space
                y_r1_c, theta_r1_c = trace_ray(
                    y_obj_c, theta_obj_c, f_r1, f_r1
                )
                y_r2_c, theta_r2_c = trace_ray(
                    y_r1_c, theta_r1_c, f_r1 + f_r2, f_r2
                )

                # Reject systems exceeding collimated beam angle limits
                if (
                    np.abs(theta_obj_c) > max_angle_rad
                    or np.abs(theta_r2_c) > max_angle_rad
                ):
                    rejection_reasons["Chief ray angle limit exceeded"] += 1
                    continue

                # Trace upper marginal ray across the full optical path
                y_obj_top, th_obj_top = trace_ray(
                    y_edge_fov, theta_marginal, f_obj, f_obj
                )
                y_r1_top, th_r1_top = trace_ray(
                    y_obj_top, th_obj_top, f_r1, f_r1
                )
                y_r2_top, th_r2_top = trace_ray(
                    y_r1_top, th_r1_top, f_r1 + f_r2, f_r2
                )
                y_tube_top, _ = trace_ray(y_r2_top, th_r2_top, f_r2, f_tube)

                # Trace lower marginal ray across the full optical path
                y_obj_bot, th_obj_bot = trace_ray(
                    y_edge_fov, -theta_marginal, f_obj, f_obj
                )
                y_r1_bot, th_r1_bot = trace_ray(
                    y_obj_bot, th_obj_bot, f_r1, f_r1
                )
                y_r2_bot, th_r2_bot = trace_ray(
                    y_r1_bot, th_r1_bot, f_r1 + f_r2, f_r2
                )
                y_tube_bot, _ = trace_ray(y_r2_bot, th_r2_bot, f_r2, f_tube)

                # Calculate required beam diameters at each lens surface
                d_real_r1 = 2.0 * max(np.abs(y_r1_top), np.abs(y_r1_bot))
                d_real_r2 = 2.0 * max(np.abs(y_r2_top), np.abs(y_r2_bot))
                d_real_tube = 2.0 * max(np.abs(y_tube_top), np.abs(y_tube_bot))

                # Verify procurement availability and clear aperture requirements
                avail_r1, size_r1 = check_lens_procurement(
                    f_r1, d_real_r1, allowed_footprint_m, catalog_dict
                )
                avail_r2, size_r2 = check_lens_procurement(
                    f_r2, d_real_r2, allowed_footprint_m, catalog_dict
                )
                avail_tube, size_tube = check_lens_procurement(
                    f_tube, d_real_tube, allowed_footprint_m, catalog_dict
                )

                if not (avail_r1 and avail_r2 and avail_tube):
                    rejection_reasons[
                        "Commercial lens availability or clear aperture limit reached"
                    ] += 1
                    continue

                # Compute clear aperture fill factors relative to outer diameter
                util_r1 = d_real_r1 / physical_diameters_m[size_r1]
                util_r2 = d_real_r2 / physical_diameters_m[size_r2]
                util_tube = d_real_tube / physical_diameters_m[size_tube]
                max_fill_factor_pct = (
                    max(util_r1, util_r2, util_tube) * 100.0
                )

                # Sum commercial unit prices for the three-lens line
                total_price = (
                    get_lens_price(specs_dict, size_r1, f_r1)
                    + get_lens_price(specs_dict, size_r2, f_r2)
                    + get_lens_price(specs_dict, size_tube, f_tube)
                )

                # Compute deviation from target magnification
                m_error = (
                    np.abs(m_total - m_target) if m_target is not None else 0.0
                )

                # Store design metrics for viable optical layout
                valid_combinations.append({
                    "f_relay1 (mm)": f_r1 * 1e3,
                    "f_relay2 (mm)": f_r2 * 1e3,
                    "f_tube (mm)": f_tube * 1e3,
                    "M_relay": m_relay,
                    "M_total": m_total,
                    "Relay 1 diam": size_r1,
                    "Relay 2 diam": size_r2,
                    "Tube lens diam": size_tube,
                    "Track length (cm)": l_track * 100.0,
                    "Max chief ray angle (deg)": np.rad2deg(
                        max(np.abs(theta_obj_c), np.abs(theta_r2_c))
                    ),
                    "Max fill factor (%)": max_fill_factor_pct,
                    "Total price (€)": total_price,
                    "Target deviation": m_error,
                })

    # Convert valid design collection to pandas DataFrame
    df_results = pd.DataFrame(valid_combinations)

    # Print evaluation metrics and rejection breakdown
    print(f"Total configurations evaluated: {total_evaluated}")
    print("\nRejection breakdown by filtering criteria (ranked by impact):")

    for reason, count in rejection_reasons.most_common():
        percentage = (
            (count / total_evaluated) * 100.0 if total_evaluated > 0 else 0.0
        )
        print(
            f"  - {reason}: {count} configurations rejected ({percentage:.2f}%)"
        )

    # Sort results by target deviation if applicable and return DataFrame
    if not df_results.empty:
        if m_target is not None:
            df_results = df_results.sort_values(
                by="Target deviation"
            ).drop(columns=["Target deviation"])
        else:
            df_results = df_results.drop(
                columns=["Target deviation"], errors="ignore"
            )
        return df_results.reset_index(drop=True)

    return pd.DataFrame()
```

### Verification of candidates via ray tracing.

Once the brute-force optimization filter restricts the commercial catalog to combinations that simultaneously satisfy the track length constraints and structural clear apertures, a high-fidelity paraxial verification is required. While nominal algebraic scaling bounds the search space, it treats the optics as idealized transfer metrics. To guarantee that the peripheral field rays do not undergo subtle clipping or severe geometric deviations, we map the exact trajectory of both on-axis and off-axis fluorescence bundles throughout the entire multi-stage track.

This is achieved by implementing a strict thin-lens ABCD matrix propagation sequence. The framework initializes two boundary configurations at the atomic object plane ($z=0$): a central fluorescence bundle defined by the objective's physical numerical aperture $\left(\pm\text{NA}\right)$, and an off-axis bundle originating from the maximum edge of the physical field of view $\left(\pm\frac{\text{FoV}}2\right)$. By tracking the continuous spatial evolution of these ray envelopes up to the qCMOS sensor plane, we visually confirm the absence of clipping across downstream apertures and ensure that the strict geometric field magnification $\left(M_{\text{field}} = \frac{f_{\text{tube}}}{f_{\text{obj}} \cdot M_{\text{relay}}}\right)$ accurately focuses the edge-most atoms within the active vertical height of the Hamamatsu sensor.

```python
def propagate_paraxial_ray(y0, theta0, f_obj, f_r1, f_r2, f_tube):
    """
    Propagates a single paraxial ray sequentially through the pupil conjugation system.
    Follows exact textbook ABCD matrix multiplication for thin lenses and free spaces.
    All variables and constants are handled in millimeters.
    """
    # Initialize array to store ray height (y) and angle (theta) at each optical plane
    y = np.zeros(6)
    theta = np.zeros(6)
    
    # Boundary conditions at the object plane (Atoms at z = 0)
    y[0] = y0
    theta[0] = theta0
    
    # 1. Propagate over distance f_obj to Objective, then refract at the objective lens
    y[1] = y[0] + theta[0] * f_obj
    theta[1] = theta[0] - (y[1] / f_obj)
    
    # 2. Propagate over distance f_r1 to Relay 1, then refract at Relay 1 lens
    y[2] = y[1] + theta[1] * f_r1
    theta[2] = theta[1] - (y[2] / f_r1)
    
    # 3. Propagate over distance (f_r1 + f_r2) to Relay 2, then refract at Relay 2 lens
    y[3] = y[2] + theta[2] * (f_r1 + f_r2)
    theta[3] = theta[2] - (y[3] / f_r2)
    
    # 4. Propagate over distance f_r2 to Tube Lens, then refract at Tube Lens
    y[4] = y[3] + theta[3] * f_r2
    theta[4] = theta[3] - (y[4] / f_tube)
    
    # 5. Propagate final distance f_tube to qCMOS Sensor plane
    y[5] = y[4] + theta[4] * f_tube
    theta[5] = theta[4]
    
    return y


def plot_paraxial_ray_tracing(config_row,
                            objective_efl=OBJECTIVE_EFL,
                            objective_na=OBJECTIVE_NA, 
                            max_physical_fov=MAX_PHYSICAL_FOV,
                            lens_type="Achromatic doublets",
                            fill_factor_limit_singlets=.3,
                            fill_factor_limit_doublets=.5,
                            fill_factor_limit_aspherics=.9
                            ):
    """
    Plots the paraxial ray tracing for the fluorescence line using the corrected propagation helper.
    """
    # Standardize all internal dimensions to millimeters to match catalog data
    f_obj = objective_efl * 1e3
    f_r1 = config_row["f_relay1 (mm)"]
    f_r2 = config_row["f_relay2 (mm)"]
    f_tube = config_row["f_tube (mm)"]
    
    # Mechanical diameter mapping from commercial catalog specifications
    size_mapping = {
            "0.5\"": 12.7, 
            "1.0\"": 25.4, 
            "1.5\"": 38.1, 
            "2.0\"": 50.8,
            "3.0\"": 76.2,
            "4.0\"": 101.6
        }
    
    # Compute effective mechanical diameters in millimeters for clear aperture checks
    d_obj = 2.0 * objective_na * f_obj 
    d_r1 = size_mapping.get(config_row["Relay 1 diam"], 25.4)
    d_r2 = size_mapping.get(config_row["Relay 2 diam"], 25.4)
    d_tube = size_mapping.get(config_row["Tube lens diam"], 25.4)
    
    # Construct physical coordinates along the optical axis for plotting (in cm)
    # Note: distance between the second relay lens and tube lens is strictly f_r2
    z_atoms = 0.0
    z_obj = f_obj
    z_r1 = z_obj + f_r1
    z_r2 = z_r1 + f_r1 + f_r2
    z_tl = z_r2 + f_r2
    z_sensor = z_tl + f_tube

    # Convert optical axis positions from millimeters to centimeters for plotting
    z_planes_cm = np.array([z_atoms, z_obj, z_r1, z_r2, z_tl, z_sensor]) / 10.0
    
    # Correct angular boundaries based on physical numerical aperture
    theta_marginal = np.arcsin(objective_na)
    y_edge = (max_physical_fov * 1e3) / 2.0
    
    # Model ray trajectories using the corrected physical increment logic
    # Calculate central atom (on-axis) ray trajectories
    y_center_top = propagate_paraxial_ray(0.0, theta_marginal, f_obj, f_r1, f_r2, f_tube)
    y_center_bot = propagate_paraxial_ray(0.0, -theta_marginal, f_obj, f_r1, f_r2, f_tube)
    y_center_chief = propagate_paraxial_ray(0.0, 0.0, f_obj, f_r1, f_r2, f_tube)
    
    # Calculate field edge (off-axis) ray trajectories
    y_edge_chief = propagate_paraxial_ray(y_edge, 0.0, f_obj, f_r1, f_r2, f_tube)
    y_edge_top = propagate_paraxial_ray(y_edge, theta_marginal, f_obj, f_r1, f_r2, f_tube)
    y_edge_bot = propagate_paraxial_ray(y_edge, -theta_marginal, f_obj, f_r1, f_r2, f_tube)
    
    # Matplotlib canvas configuration
    fig, ax = plt.subplots(figsize=(11, 5.5))
    
    # Render central atom on-axis fluorescence bundles
    ax.fill_between(z_planes_cm, y_center_top, y_center_bot, color="#912491", alpha=0.15, label="On-axis bundle")
    ax.plot(z_planes_cm, y_center_top, color="#4A154B", lw=1.5, ls="-")
    ax.plot(z_planes_cm, y_center_bot, color="#4A154B", lw=1.5, ls="-")
    ax.plot(z_planes_cm, y_center_chief, color="#4A154B", lw=1.2, ls="--")
    
    # Render off-axis field edge fluorescence bundles
    ax.fill_between(z_planes_cm, y_edge_top, y_edge_bot, color="#00BCD4", alpha=0.15, label="Off-axis bundle")
    ax.plot(z_planes_cm, y_edge_top, color="#00838F", lw=1.5, ls="-")
    ax.plot(z_planes_cm, y_edge_bot, color="#00838F", lw=1.5, ls="-")
    ax.plot(z_planes_cm, y_edge_chief, color="#00838F", lw=1.2, ls="--")
    
    # Display optical elements with real physical diameters and catalog focal lengths
    lens_positions_cm = [z_obj / 10.0, z_r1 / 10.0, z_r2 / 10.0, z_tl / 10.0]
    lens_diameters = [d_obj + 5, d_r1, d_r2, d_tube]
    lens_labels = ["Obj", "Relay 1", "Relay 2", "Tube lens"]
    
    # Pack focals to easily print them in the loop (Objective is fixed)
    lens_focals = [f_obj, f_r1, f_r2, f_tube]
    
    # Size names for the catalog labels
    lens_sizes_inch = [
        "N/A", 
        config_row["Relay 1 diam"], 
        config_row["Relay 2 diam"], 
        config_row["Tube lens diam"]
    ]
    
    # Determine the structural filling factor limits depending on active lens system type
    if lens_type == "singlet":
        fill_factor = fill_factor_limit_singlets
    elif lens_type == "doublet":
        fill_factor = fill_factor_limit_doublets
    else:
        fill_factor = fill_factor_limit_aspherics
    
    # Draw physical lens boundaries and mechanical limits on canvas
    for z_pos, diam, name, focal, size_in in zip(lens_positions_cm, lens_diameters, lens_labels, lens_focals, lens_sizes_inch):
            ax.plot([z_pos, z_pos], [-diam / 2.0, diam / 2.0], color="#212121", lw=3.5, zorder=4)
            
            # Draw geometrical limits for commercial catalog elements (excluding custom objective housing)
            if name != "Obj":
                # Compute clear aperture envelope and safety boundaries
                ca_height = (diam / 2.0) * 0.9
                fill_height = (diam / 2.0) * fill_factor
                
                # Plot physical markers for clear aperture thresholds
                ax.plot([z_pos - 0.4, z_pos + 0.4], [ca_height, ca_height], color="k", lw=2, zorder=5)
                ax.plot([z_pos - 0.4, z_pos + 0.4], [-ca_height, -ca_height], color="k", lw=2, zorder=5)
                
                # Plot physical markers for filling factor limitations
                ax.plot([z_pos - 0.6, z_pos + 0.6], [fill_height, fill_height], color="k", lw=2.5, zorder=5)
                ax.plot([z_pos - 0.6, z_pos + 0.6], [-fill_height, -fill_height], color="k", lw=2.5, zorder=5)
            
            # Build a structured multi-line label for clean vertical stacking
            if name == "Obj":
                label_text = f"{name}\nf={focal:.1f}mm"
            else:
                label_text = f"{name}\nf={focal:.0f}mm\nØ={size_in}"
                
            ax.text(z_pos, diam / 2.0 + 1.0, label_text, ha="center", va="bottom", fontsize=15, weight="bold", color="#212121")
        
    # Sensor label update to also show its physical height clear limits
    ax.plot([z_sensor / 10.0, z_sensor / 10.0], [-SENSOR_HEIGHT * 1e3 / 2.0, SENSOR_HEIGHT * 1e3 / 2.0], color="#212121", lw=3.5, zorder=4)
    ax.text(z_sensor / 10.0+1, SENSOR_HEIGHT * 1e3 / 2.0 + 1.0, f"qCMOS sensor\nH={SENSOR_HEIGHT*1e3:.2f}mm", ha="center",
            va="bottom", fontsize=15, weight="bold", color="#212121")

    # Draw lower dimension indicators for inter-element distances
    max_aperture = max(lens_diameters)
    y_cota = -max_aperture / 2.0 - 2.5
    
    # Complete coordinates including initial (Atoms) and final (Sensor) planes
    all_planes_cm = [z_atoms / 10.0, z_obj / 10.0, z_r1 / 10.0, z_r2 / 10.0, z_tl / 10.0, z_sensor / 10.0]
    
    # Effective half-heights for each vertical plane to extend dotted reference lines appropriately
    plane_half_heights = [
        0.0,                                 # Atoms (point source at y=0)
        lens_diameters[0] / 2.0,              # Objective
        lens_diameters[1] / 2.0,              # Relay 1
        lens_diameters[2] / 2.0,              # Relay 2
        lens_diameters[3] / 2.0,              # Tube lens
        (SENSOR_HEIGHT * 1e3) / 2.0           # Sensor
    ]
    
    # Render dashed vertical reference lines and distance indicators along optical axis
    for i in range(len(all_planes_cm) - 1):
        z_start = all_planes_cm[i]
        z_end = all_planes_cm[i+1]
        dist_mm = (z_end - z_start) * 10.0
        
        h_start = plane_half_heights[i]
        h_end = plane_half_heights[i+1]
        
        # Fine dotted vertical lines pointing down from each physical surface
        if i == 0:
            ax.plot([z_start, z_start], [0.0, y_cota - 0.5], color="#4A154B", ls=":", lw=0.8, alpha=0.6)
        else:
            ax.plot([z_start, z_start], [-h_start, y_cota - 0.5], color="#4A154B", ls=":", lw=0.8, alpha=0.6)
            
        if i == len(all_planes_cm) - 2:
            ax.plot([z_end, z_end], [-h_end, y_cota - 0.5], color="#4A154B", ls=":", lw=0.8, alpha=0.6)
            
        # Bidirectional arrow and distance text with clean vertical spacing
        ax.annotate('', xy=(z_start, y_cota), xytext=(z_end, y_cota),
                    arrowprops=dict(arrowstyle="<->", color="#4A154B", lw=1.2, shrinkA=0, shrinkB=0))
        ax.text((z_start + z_end) / 2.0, y_cota + 0.8, f"{dist_mm:.1f} mm", ha="center", va="bottom", fontsize=11, weight="bold", color="#4A154B")

    # Final baseline plot to close vertical reference line for sensor
    ax.plot([all_planes_cm[-1], all_planes_cm[-1]], [-plane_half_heights[-1], y_cota - 0.5], color="#4A154B", ls=":", lw=0.8, alpha=0.6)

    # Window aesthetics and labeling
    ax.set_xlabel(r"$\mathbf{Optical\ axis\ position\ (z)\ [cm]}$", fontsize=18, color="#4A154B", weight="bold")
    ax.set_ylabel(r"$\mathbf{Ray\ height\ (y)\ [mm]}$", fontsize=18, color="#4A154B", weight="bold")
    ax.axhline(0, color="#b0a8b9", ls=":", alpha=0.6)
    
    ax.set_xlim(-3.0, (z_sensor / 10.0) + 7)
    ax.set_ylim(-max_aperture / 2.0 - 5.0, max_aperture / 2.0 + (max_aperture * 0.25))
    ax.grid(True, which="both", linestyle="--", alpha=0.3, color="#b0a8b9")
    
    ax.tick_params(colors="#4A154B", labelsize=18, width=1.5)
    for spine in ax.spines.values():
        spine.set_edgecolor("#4A154B")
        spine.set_linewidth(1.5)
        
    for label in ax.get_xticklabels() + ax.get_yticklabels():
        label.set_weight('bold')
        
    ax.legend(fontsize=13, prop={'weight':'bold', 'size':13}, facecolor='#f4f0fa', edgecolor='#4A154B', loc='upper right')
    total_price = config_row["Total price (€)"]
    total_mag = config_row["M_total"]
    title_text = f"{lens_type} | Price: {total_price:.2f} € | M = {total_mag:.2f}"
    
    ax.set_title(title_text, fontsize=18, color="#4A154B", weight="bold", pad=20)
    plt.tight_layout()
    plt.show()
```
